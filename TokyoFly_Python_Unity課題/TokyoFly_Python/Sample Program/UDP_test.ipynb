```python
from socket import *

class udp_server:
    def __init__(self):
        # 送信側IP
        SrcIP = "127.0.0.1"    # localhost
        # 送信側ポート番号
        SrcPort = 11111
        # 送信側アドレスをtupleに格納
        self.SrcAddr = (SrcIP, SrcPort)

        # 受信側アドレスの設定
        # 受信側IP
        DstIP = "127.0.0.1"
        # 受信側ポート番号
        DstPort = 22222
        # 受信側アドレスをtupleに格納
        self.DstAddr = (DstIP, DstPort)
        
        # 受信時のバッファサイズ
        self.buffer_size = 4092

        # ソケット作成
        self.udpClntSock = socket(AF_INET, SOCK_DGRAM)
        # 送信側アドレスでソケットを設定
        self.udpClntSock.bind(self.SrcAddr)
        
    
    def __delete__(self):
        del self.udpClntSock
        
        
    def sendString(self, data):
        # バイナリに変換
        data = data.encode('utf-8')

        # 受信側アドレスに送信
        self.udpClntSock.sendto(data, self.DstAddr)
        
    def sendInt(self, intVal):
        data = intVal.to_bytes(4, 'little')
        self.udpClntSock.sendto(data, self.DstAddr)
        
    def receive(self):
        data = self.udpClntSock.recv(self.buffer_size)
        return data
```

```python
udp_sample = udp_server()
```

```python
while (True):
    data = udp_sample.receive()
    intVal = int.from_bytes(data, 'little')
    print(intVal)
    udp_sample.sendInt(intVal)
```

```python
import struct
while (True):
    data = udp_sample.receive()
    floatVal = struct.unpack('<ff', data)
    print(floatVal)
```

```python
del udp_sample
```
